Que . 76 .  Write a Python program to read a file line by line and store it into a list

```python
def file_to_list(filename):
    lines = []
    with open(filename , 'r') as f:
        n = f.readlines()
        for i in n:
            line = i
            lines.append(line.strip())
    print(lines)

file_to_list('test.txt')
```

Que . 78 . Write a python program to find the longest words.

```python
def Longest_word(filename):
    with open(filename, 'r') as f:
        file_str = f.read()
    
    words = file_str.split()
    max_len = 0
    ans = ''
    print(words)
    for word in words:
        len_word = len(word)
        if len_word > max_len:
            ans = word
            max_len = len_word
    return f'{ans} is max length with {max_len} chars'

print(Longest_word('test.txt'))
```

Que . 80 . Write a Python program to count the frequency of words in a file. 

```python
def freq_words(filename):
    with open(filename, 'r') as f:
        file_str = f.read()

    words = file_str.split()

    words_and_freq = {}
    for word in words:
        word = word.strip()  
        if word in words_and_freq:
            words_and_freq[word] += 1
        else:
            words_and_freq[word] = 1
        
    print(words_and_freq)


freq_words('test.txt')
```

Que . 82 . Write a Python program to copy the contents of a file to another file. 

```python
def copy_file(filename , copyFilename):
    with open(filename , 'r') as f :
        lines = f.readlines()
    with open(copyFilename , 'w+') as copy_f:
        copy_f.writelines(lines)
    with open(copyFilename , 'r') as f :
        data_copied = f.read()
        print(data_copied)
copy_file('test2.txt' , 'test2_copy.txt')
```

Que . 83 . Explain Exception handling? What is an Error in Python? 

<h4>Exception  handling </h4> in python is mechanism that helps prevent or manage run time errors that can 
potentially stop the flow of the program/code .And help getting them managed using raise keyword to raise an exception.

try :    In this lock we write code that might raise an exception or create error

except : These blocks follow the try block and specify how to handle specific types of exceptions. 
         If an exception occurs within the try block, Python searches for a matching except block to execute.

else :   This block is excecuted when no exception is excecuted .

finally :This block always executes, regardless of whether an exception occurred or was handled.
         It's typically used for cleanup operations, such as closing files or releasing resources.

<h4>Error in Python </h4>
An error is an issue in a program that prevents the program from completing its task. In comparison, 
an exception is a condition that interrupts the normal flow of the program. Both errors and exceptions are a type of runtime error, 
which means they occur during the execution of a program.

Errors are mainly 3 types that is logical error , syntax error and runtime errors(exception)

Que . 89 . How Do You Handle Exceptions with Try/Except/Finally in Python? 
Explain with coding snnippet

```python
def divide(x, y): 
    try: 
        result = x // y 
    except ZeroDivisionError: 
        print("Sorry ! You are dividing by zero ") 
    else:
        print("Yeah ! Your answer is :", result) 
    finally:  
        # this block is always executed   
        # regardless of exception generation.  
        print('This is always executed')   

# Look at parameters and note the working of Program 
divide(3, 2) 
divide(3, 0)
```
