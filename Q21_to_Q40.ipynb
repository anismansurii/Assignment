<b><h3>Que 22 Write a Python function to reverses a string if its length is a multiple 
of 4.</h3></b>

```python
def reverse_str(str):
    print(len(str))
    if(len(str)%4==0):
        char_words = []
        for i in str:
            char_words.append(i)
        char_words = char_words[::-1]
        reversed_str = "".join(char_words)
        print(reversed_str)
    else:
        print("String length is not multiple of 4")
reverse_str("abcd")
```

<b><h3> Que . 28 .  Differentiate between append () and extend () methods?  </b></h3>

append(element): This method adds a single element to the end of the list. If the element you are appending is an 
iterable (like another list or a tuple), it will be added as a single item at the end of the list, maintaining its structure.

extend(iterable): This method adds multiple elements from an iterable (like another list, tuple, string, or set) to
 the end of the list. It iterates over the elements of the provided iterable and adds each individual element to the list.


<b><h3> Que . 29 . Write a Python function to get the largest number, smallest num 
and sum of all from a list.

```python
import sys
def list_max_min_sum(li):
    max = -1
    min = sys.maxsize
    sum = 0
    for i in li:
        if i > max:
            max = i
        if i < min :
            min = i
        sum+=i
    print(f"Sum of List elements {li} is {sum}")
    print(f"Max of List elements {li} is {max}")
    print(f"Min of List elements {li} is {min}")

list_max_min_sum([1,2,3,4,5])
```
