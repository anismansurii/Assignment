Que . 61 . Write a Python function to calculate the factorial of a number 
Note : a nonnegative integer

```python
def Fact_number(num):
    if num == 0 or num == 1:
        return 1
    elif num < 0:
        return -1
    else:
        return Fact_number(num-1) * num
print(Fact_number(5))
```

Que . 64 . Write a Python function that checks whether a passed string is 
palindrome or not 

```python
def palindrome_check(str):
    new_str = ''
    n = len(str)
    for i in range(0,n):
        new_str += str[n-1-i]
    
    if new_str==str:
        return f"Yes {str} is palindrome"
    else:
        return False
palindrome_check('abcba')
```

<b>Que . 65 . How Many Basic Types of Functions Are Available in Python? </b>

        -> Built-in Functions:

        These functions are pre-defined and readily available within the Python interpreter. They perform common tasks and are always accessible without requiring any explicit import statements.
        Examples: print(), len(), type(), max(), min(), sum(), int(), str(), float(). 

        -> User-defined Functions:

        These are functions created by the user to perform specific, custom tasks. They are defined using the def keyword.
        Example: 
                 def add(a,b):
                    return a+b
                 print(add(1,2)) # Output will be 3

        -> Anonymous Functions (Lambda Functions):

        These are small, single-expression functions defined without a name using the lambda keyword. They are typically used for quick, inline operations.
        Example: add = lambda x, y: x + y

        -> Recursive Functions:

        Functions that call themselves to solve a problem by breaking it down into smaller, similar sub-problems.
        Example: A factorial calculation function. 
                see question number 61 

        -> Generator Functions:

        Functions that use the yield keyword to produce a sequence of values iteratively, rather than returning a single value.
        They are memory-efficient for large sequences. 

Que . 66 . How can you pick a random item from a list or tuple? 

Ans   ->   To pick a random item from a list or tuple in Python, the random.choice() function from the random module is used.  

```python
#Example 
import random

# Example with a list
my_list = ["apple", "banana", "cherry", "date"]
random_item_from_list = random.choice(my_list)
print(f"Random item from list: {random_item_from_list}")

# Example with a tuple
my_tuple = ("red", "green", "blue", "yellow")
random_item_from_tuple = random.choice(my_tuple)
print(f"Random item from tuple: {random_item_from_tuple}")
```

Que . 68 . How can you get a random number in python? 

Ans  ->   1. Generating Random Integers: random.randint(a, b): 
             This function returns a random integer N such that a <= N <= b (inclusive of both endpoints).

             Ex :    import random
                     random_integer = random.randint(1, 10) # Generates a random integer between 1 and 10 (inclusive)
                     print(random_integer)

          2. random.randrange(start, stop, step): This function returns a random integer from the range specified 
             by start (inclusive) and stop (exclusive), with an optional step value.

             Ex :       import random
                        random_integer_range = random.randrange(0, 10, 2) 
                        # Generates a random even integer between 0 and 8 (exclusive of 10)
                        print(random_integer_range)

          3 . Generating Random Floating-Point Numbers:
              random.random(): This function returns a random floating-point number in the range (0.0, 1.0) 
                               (inclusive of 0.0, exclusive of 1.0).

          4. random.uniform(a, b): This function returns a random floating-point number N.
             such that a <= N <= b or b <= N <= a (inclusive of both endpoints).

          5 . Shuffling a Sequence:
              random.shuffle(x): This function shuffles the elements of a sequence x in place (modifies the original sequence).

              Ex :     import random
                       numbers = [1, 2, 3, 4, 5]
                        random.shuffle(numbers)
                       print(numbers)

```python
import random


random_integer = random.randint(1, 10) # Generates a random integer between 1 and 10 (inclusive)
print("Using random.randint",random_integer)


random_integer_range = random.randrange(0, 10, 2) 
# Generates a random even integer between 0 and 8 (exclusive of 10)
print("Using random.randrange",random_integer_range)


random_float = random.random()
print("Using random.random",random_float)


random_float_range = random.uniform(1.5, 5.5) # a and b both included
print("Using random.uniform",random_float_range)

    
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print("Randomly sequenced numbers ",numbers)
```

Que . 71 . What is File function in python? What are keywords to create 
and write file. 

In Python, the term "file function" generally refers to the built-in functions and methods used for interacting with files on your system. 
The primary function for file handling is open(), which returns a file object. 
This file object then provides various methods for reading from or writing to the file.

1. open() function:

This is the core function for opening a file.
It takes the filename and an optional mode as arguments.
Modes include:
'r' (read - default)
'w' (write - overwrites existing file or creates a new one)
'a' (append - adds to the end of the file)
'x' (create - creates a new file, raises error if it exists)
'b' (binary mode)
't' (text mode - default)

2. File object methods:

Once a file is opened, the returned file object provides methods for operations.
read(): Reads the entire content of the file or a specified number of characters.
readline(): Reads a single line from the file.
readlines(): Reads all lines from the file and returns them as a list of strings.
write(): Writes a string to the file.
writelines(): Writes a list of strings to the file.
close(): Closes the file, releasing system resources. This is crucial for ensuring data integrity.

3. with statement:

The with statement is the recommended way to handle files in Python.
It ensures that the file is automatically closed, even if errors occur.

Que . 75 .  Write a Python program to read last n lines of a file. 

```python
def read_last_line(file_name , n):
    n=4
    lines= []
    with open(file_name , 'r') as f:
        for i in range(0,len(file_name)):
            line = f.readline()
            lines.append(line.strip())

    for i in range(n , len(lines)):
        print(lines[i])


read_last_line('test1.txt' , 4)
```
