<b><h3>Que . 7 . How memory is managed in Python? </h3></b>
 
Ans :  Python automatically handles memory for you, unlike some languages where you have to do it manually. It does this using
       a special area called a "heap" to store all your data and objects. Python tracks how many references (or 'names' in Python)
       point to each object using a technique called reference counting. When an object has no more references, it's considered unused,
       and Python's garbage collector automatically frees up that memory space. The garbage collector is also smart enough to detect 
       complex loops where objects might reference each other, preventing them from being cleaned up by simple reference counting. 
       This automatic process allows programmers to focus on writing code rather than worrying about memory details. 


<b><h3>Que . 8 .  What is the purpose continuing statement in python? </h3></b>

Ans :     The continue statement in Python is used within loops to skip the current iteration and move to the next one.
          When encountered, it stops the execution of the remaining code within that specific iteration's loop body and 
          jumps directly to the beginning of the next iteration. This is useful for situations where you want to ignore 
          certain values or conditions within a loop, without exiting the entire loop prematurely.

        Python uses zero-based indexing, meaning the first element is at index 0, the second at index 1, and so on. 


        Python offers a convenient feature called negative indexing, allowing you to access elements from the end of a sequence.
        The last element has an index of -1.

        ex :   H   e   l   l  o
              -5  -4  -3  -2  -1

        Negative indexes can be combined with slicing to extract portions of a sequence from the end.
