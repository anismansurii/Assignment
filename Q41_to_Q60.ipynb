Que . 46 Write a Python program to convert a list of tuples into a dictionary. 
 

```python
tpl_list = (["John" , "America" , "1234567890"],
            ["Alex" , "Australia" , "1234565432"],
            ["Bob" , "UK" , "1233211234"]
            )
dict_ppl ={}
count=0
for i in tpl_list:
    name = f"name_{count}"
    country = f"Country_{count}"
    phone = f"Phone_number_{count}"
    dict_ppl[name] = i[0]
    dict_ppl[country] = i[1]
    dict_ppl[phone] = i[2]
    count+=1

for key , value in dict_ppl.items():
    print(f"Key is {key} and value is {value}")
    
```

Que . 51 . How Do You Traverse Through a Dictionary Object in Python 

Ans : We can traverse through dict using dict.keys , dict.values , dict.items 
      using this dict methods dict.keys we can iterate through keys of dict and 
      using dict.values we can iterate over a values of dict not key 
      and at the last we can use dict.items which provides the access to both dict keys and values 
      example is given below

```python
dict_info = {"Name":"Alice" , "Add":"India" , "Email":"Alice@gmail.com"}

for i in dict_info.keys():
    print(i)
for i in dict_info.values():
    print(i)
for i in dict_info.items():
    print(i)
```

Que . 58 .  Write a Python program to combine values in python list of dictionaries. 
Sample data: [{'item': 'item1', 'amount': 400}, {'item': 'item2', 'amount': 
300}, o {'item': 'item1', 'amount': 750}] 

Expected Output: 
• Counter ({'item1': 1150, 'item2': 300}) 

```python
list_dict =  [
              {'item': 'item1', 'amount': 400}, 
              {'item': 'item2', 'amount': 300}, 
              {'item': 'item1', 'amount': 750}
              ]
dict_ans = {}
for dict_n in list_dict:
    if dict_n['item'] in dict_ans:
          dict_ans[dict_n['item']] += dict_n['amount']

    else:
         dict_ans[dict_n['item']] = dict_n['amount']
dict_ans
```

Que . 59 . Write a Python program to create a dictionary from a string. 
Note: Track the count of the letters from the string. 

```python
str_dict = 'Banana is fruit and so apple is'
str_dict = str_dict.replace(" " , "")

dict_chars = {}

char_words = []
for i in str_dict:
    if i == " ":
        continue
    else:
        char_words.append(i)

# print(char_words)
uniq_chars = set(char_words)

char_words = list(uniq_chars)


for i in str_dict:
    dict_chars[i] = 0

for i in str_dict:
    dict_chars[i]+=1 

print(dict_chars)

```
